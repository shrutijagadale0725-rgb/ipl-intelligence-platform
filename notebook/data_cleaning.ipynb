```python
matches.replace('Delhi Daredevils', 'Delhi Capitals', inplace=True)
matches.replace('Kings XI Punjab', 'Punjab Kings', inplace=True)
matches.replace('Royal Challengers Bangalore', 'Royal Challengers Bengaluru', inplace=True)
```

```python
#1️⃣ Toss Impact ✅
matches['toss_win_match_win'] = matches['toss_winner'] == matches['winner']
matches['toss_win_match_win'].value_counts()
```

```python
toss_impact = matches['toss_win_match_win'].value_counts(normalize=True) * 100

print(toss_impact)
```

```python
import matplotlib.pyplot as plt

toss_impact.plot(kind='bar')

plt.title("Does Winning Toss Help Win Match?")
plt.xlabel("Won Toss & Match?")
plt.ylabel("Percentage")

plt.xticks(rotation=0)

plt.show()
```

### Insight:
Teams winning the toss also win the match a significant percentage of the time, suggesting toss can influence match outcome.

```python
#2️⃣TOP RUN SCORERS ANALYSIS
top_batters = deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False)

top_batters.head(10)
```

```python
import matplotlib.pyplot as plt

top10 = top_batters.head(10)

plt.figure(figsize=(12,6))

top10.plot(kind='bar')

plt.title("Top 10 IPL Run Scorers")
plt.xlabel("Players")
plt.ylabel("Runs")

plt.xticks(rotation=45)

plt.show()
```

### Insight:
Virat Kohli and other top-order batters dominate IPL run-scoring charts due to consistency across multiple seasons.

```python
#3️⃣🏏 Top Wicket Takers Analysis
wickets = deliveries[deliveries['dismissal_kind'] != 'run out']
```

```python
top_bowlers = wickets.groupby('bowler')['is_wicket'].sum().sort_values(ascending=False)
top_bowlers.head(10)
```

```python
top10_bowlers = top_bowlers.head(10)

plt.figure(figsize=(12,6))

top10_bowlers.plot(kind='bar', color='orange')

plt.title("Top 10 IPL Wicket Takers")
plt.xlabel("Bowlers")
plt.ylabel("Wickets")

plt.xticks(rotation=45)

plt.show()
```

### Insight:
Experienced death-over and powerplay bowlers dominate IPL wicket charts due to higher wicket-taking opportunities.

```python
#4️⃣ Chasing vs Defending ✅
matches['result'].value_counts()
```

```python
chasing_wins = matches[matches['result'] == 'wickets'].shape[0]

defending_wins = matches[matches['result'] == 'runs'].shape[0]

print("Chasing Wins:", chasing_wins)
print("Defending Wins:", defending_wins)
```

```python
labels = ['Chasing Wins', 'Defending Wins']
sizes = [chasing_wins, defending_wins]

plt.figure(figsize=(7,7))

plt.pie(sizes, labels=labels, autopct='%1.1f%%')

plt.title("Chasing vs Defending Wins in IPL")

plt.show()
```

### Insight:
Teams batting second have won a significant percentage of IPL matches, suggesting chasing is often advantageous in T20 cricket.

```python
#5️⃣ Venue Analysis ✅
chasing_matches = matches[matches['result'] == 'wickets']
```

```python
venue_chasing = chasing_matches['venue'].value_counts()
```

```python
venue_chasing.head(10)
```

```python
top_venues = venue_chasing.head(10)

plt.figure(figsize=(12,6))

top_venues.plot(kind='bar', color='green')

plt.title("Top IPL Venues Favoring Chasing")
plt.xlabel("Venue")
plt.ylabel("Successful Chases")

plt.xticks(rotation=75)

plt.show()
```

### Insight:
Certain IPL venues show a higher number of successful chases, suggesting pitch conditions and dew factors may favor batting second.

```python
#6️⃣ Season Trends Analysis
season_runs = deliveries.groupby('match_id')['total_runs'].sum().reset_index()
```

```python
season_runs = season_runs.merge(matches[['id', 'season']],
                                left_on='match_id',
                                right_on='id')
```

```python
avg_runs_per_season = season_runs.groupby('season')['total_runs'].mean()
```

```python
plt.figure(figsize=(12,6))

avg_runs_per_season.plot(kind='line', marker='o')

plt.title("Average IPL Match Scores by Season")
plt.xlabel("Season")
plt.ylabel("Average Runs")

plt.xticks(rotation=45)

plt.show()
```

### Insight:
IPL scoring trends have generally increased across seasons due to aggressive batting strategies and modern T20 evolution.

```python
#7️⃣ Strike Rate Analysis 🔥
batter_runs = deliveries.groupby('batter')['batsman_runs'].sum()
```

```python
balls_faced = deliveries.groupby('batter')['ball'].count()
```

```python
strike_rate = (batter_runs / balls_faced) * 100
```

```python
qualified_players = balls_faced[balls_faced > 500].index

strike_rate = strike_rate[qualified_players]
```

```python
top_strike_rates = strike_rate.sort_values(ascending=False)

top_strike_rates.head(10)
```

```python
plt.figure(figsize=(12,6))

top_strike_rates.head(10).plot(kind='bar', color='purple')

plt.title("Top IPL Strike Rates")
plt.xlabel("Batters")
plt.ylabel("Strike Rate")

plt.xticks(rotation=45)

plt.show()
```

### Insight:
Aggressive middle-order and finishing batters dominate strike-rate rankings due to their ability to score quickly in limited deliveries.
